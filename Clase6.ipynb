```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
k = 100  # Precio de ejercicio
S_T = np.linspace(50, 200)
```

```python
# Ganancias de posiciones short
call_payoff = np.maximum(S_T - k, 0)
put_payoff = np.maximum(k - S_T, 0)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(S_T, call_payoff, label="Short Call", color="red")
plt.plot(S_T, put_payoff, label="Short Put", color="blue")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(k, color="gray", linestyle="--", label=f"K = {k}")

plt.xlabel("Precio del subyacente S(T)")
plt.ylabel("Ganancia")
plt.title("Ganancias de una Call y Put vendidas")
plt.legend()
plt.grid(True)
plt.show()
```

### Simulación con números aleatorios comunes para comparar call y put

```python
z = np.random.normal(size=M)

def precios_opciones(ST, strikes):
    call = [np.mean(np.maximum(ST - strike, 0)) for strike in strikes]
    put = [np.mean(np.maximum(strike - ST, 0)) for strike in strikes]
    return np.array(call), np.array(put)

# 1. Precio de la opción vs precio del activo
ST_s0 = [
    s * np.exp((r - 0.5 * var**2) * T + var * np.sqrt(T) * z)
    for s in precio1
]
call_s0 = [np.mean(np.maximum(st - k, 0)) for st in ST_s0]
put_s0 = [np.mean(np.maximum(k - st, 0)) for st in ST_s0]

# 2. Precio de la opción vs strike
ST_base = S0 * np.exp((r - 0.5 * var**2) * T + var * np.sqrt(T) * z)
call_k, put_k = precios_opciones(ST_base, k1)

# 3. Precio de la opción vs volatilidad
call_vol, put_vol = [], []
for sigma in volatilidades:
    st = S0 * np.exp((r - 0.5 * sigma**2) * T + sigma * np.sqrt(T) * z)
    call_vol.append(np.mean(np.maximum(st - k, 0)))
    put_vol.append(np.mean(np.maximum(k - st, 0)))

# 4. Precio de la opción vs tiempo
call_t, put_t = [], []
for tiempo in Tiempo:
    st = S0 * np.exp((r - 0.5 * var**2) * tiempo + var * np.sqrt(tiempo) * z)
    call_t.append(np.mean(np.maximum(st - k, 0)))
    put_t.append(np.mean(np.maximum(k - st, 0)))

# 5. Precio de la opción vs rendimiento
call_r, put_r = [], []
for rendimiento in re:
    st = S0 * np.exp(
        (rendimiento - 0.5 * var**2) * T + var * np.sqrt(T) * z
    )
    call_r.append(np.mean(np.maximum(st - k, 0)))
    put_r.append(np.mean(np.maximum(k - st, 0)))

# Gráficas
fig, axes = plt.subplots(3, 2, figsize=(14, 15))
axes = axes.ravel()

graficas = [
    (precio1, call_s0, put_s0, "Precio del activo", "Precio de la opción vs precio del activo"),
    (k1, call_k, put_k, "Strike", "Precio de la opción vs strike"),
    (volatilidades, call_vol, put_vol, "Volatilidad", "Precio de la opción vs volatilidad"),
    (Tiempo, call_t, put_t, "Tiempo", "Precio de la opción vs tiempo"),
    (re, call_r, put_r, "Rendimiento", "Precio de la opción vs rendimiento"),
]

for ax, (x, call, put, xlabel, title) in zip(axes, graficas):
    ax.plot(x, call, marker="o", label="Call")
    ax.plot(x, put, marker="o", label="Put")
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Precio")
    ax.set_title(title)
    ax.grid(True)
    ax.legend()

axes[-1].axis("off")
plt.tight_layout()
plt.show()
```

## Parte 4. Interpretación

### Precio del activo

Una opción call aumenta su valor cuando aumenta el precio del activo subyacente porque su función de pago es max(S_T - K, 0). Al subir S0, toda la distribución de posibles valores futuros de S_T se desplaza hacia arriba, lo que provoca dos efectos simultáneos: aumenta la probabilidad de que la opción termine dentro del dinero (S_T > K), y cuando esto ocurre, la ganancia S_T - K es en promedio mayor. En mis resultados esto se observa con claridad: el precio de la call pasa de 1.93 (con S0 = 80) a 26.72 (con S0 = 120).

La put se comporta de manera exactamente inversa porque su función de pago es max(K - S_T, 0). Al subir el precio del activo, disminuye la probabilidad de que S_T termine por debajo de K, y cuando ocurre, la diferencia es menor. Por eso en mi tabla el precio de la put cae de 17.93 (S0 = 80) a 1.48 (S0 = 120): a mayor precio del subyacente, la put pierde valor.

### Volatilidad

Una mayor incertidumbre siempre aumenta el valor de la opción, tanto de la call como de la put, y esto se debe a la asimetría de la función de pago: al estar acotada por abajo en cero (max(·, 0)), la opción solo se beneficia de la dispersión adicional en la dirección favorable, sin sufrir pérdidas adicionales en la dirección desfavorable. Es decir, quien compra una opción se queda con el "lado bueno" de la incertidumbre. Mis resultados confirman esto: la call sube de 7.15 (σ = 10%) a 25.83 (σ = 60%), y la put sube de 2.01 a 21.75 en ese mismo rango.

### Tiempo

Más tiempo no siempre es deseable, porque compiten dos efectos. Por un lado, más tiempo implica más varianza acumulada (el término σ√T crece), lo cual favorece el valor de la opción por la misma razón que la volatilidad. Por otro lado, más tiempo también implica una mayor deriva esperada del activo bajo la tasa libre de riesgo (el término (r − 0.5σ²)T crece), lo que hace que se espere que el activo suba, y esto no favorece a la put.

En la call, ambos efectos apuntan en la misma dirección, por lo que su valor crece de forma marcada: de 4.72 (T = 0.25) a 18.13 (T = 2). En la put, el efecto de la mayor incertidumbre gana, pero se ve parcialmente contrarrestado por la deriva esperada al alza, y por eso su crecimiento es mucho más plano: de 3.49 a 7.38 en el mismo rango.

Más tiempo resulta favorable para quien compra la opción (posición larga, sea call o put), ya que tiene más oportunidades de terminar dentro del dinero sin arriesgar más que la prima pagada. Para quien vende la opción (posición corta), más tiempo es desfavorable, porque asume el riesgo de un movimiento adverso durante una ventana más amplia sin recibir compensación adicional por ello.

### Strike

El strike modifica de forma directa la probabilidad de terminar in-the-money. Para la call, subir K reduce la probabilidad de que S_T supere al strike, por lo que su valor disminuye conforme K aumenta. Para la put ocurre lo contrario: subir K aumenta la probabilidad de que S_T quede por debajo del strike, por lo que su valor aumenta. En mi tabla de la put esto se ve con claridad: de 0.73 (K = 80) a 18.19 (K = 120).

Cabe mencionar que en mi tabla de la call contra el strike los valores casi no cambiaron (11.28 a 11.16 aproximadamente). Al revisar el código detecté que en ese loop no volví a simular S_T con el S0 base para cada valor de K, sino que reutilicé el S_T de la iteración anterior. Esto se corrige recalculando S_T dentro del mismo ciclo en el que se varía K, tal como lo hice en la función reutilizable de la Parte 3.

Una opción está in-the-money cuando su ejercicio generaría una ganancia: una call lo está cuando S_T > K (el tenedor compraría el activo más barato que su precio de mercado), y una put lo está cuando S_T < K (el tenedor vendería el activo más caro que su precio de mercado).

### Comparación con el mercado
Si yo fuera inversionista, el parámetro que más me preocuparía monitorear es la volatilidad. La razón principal no es solo la magnitud del cambio en el precio, sino la naturaleza de este parámetro frente a los demás:

El precio del activo (S0) es un dato observable en tiempo real en el mercado, así que no hay incertidumbre sobre su valor actual. El strike (K) y el tiempo a vencimiento (T) son términos del contrato: se fijan al momento de comprar la opción y no cambian después. La tasa libre de riesgo (r) es relativamente estable en el corto plazo y se puede observar en el mercado de deuda. La volatilidad (σ), en cambio, es el único parámetro que no se observa directamente: se estima a partir de datos históricos o se infiere del precio de mercado de otras opciones (volatilidad implícita), y cambia constantemente con las condiciones del mercado.

Además, como se observa en mis resultados, un aumento en σ eleva el precio tanto de la call como de la put al mismo tiempo, mientras que otros parámetros como S0 mueven ambas opciones en direcciones opuestas. Esto convierte a la volatilidad en la fuente de riesgo más relevante para quien mantiene posiciones en opciones: un error en su estimación afecta la valuación de todo un portafolio de opciones en la misma dirección, sin que un movimiento del subyacente lo compense.
