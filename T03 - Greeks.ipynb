### **Parte 2: Escenarios de mercado** *(sin simular aún)*

#### **Escenario A:** alza gradual en los próximos meses, volatilidad sin cambio  
Compraría B. Tiene Delta 0.50 y vence en 3 meses, que cubre el horizonte de próximos meses. Su Theta relativa (0.375 % diario) es moderada. Como no cambia la volatilidad, su Vega (0.25) es irrelevante, y no hay ventaja en pagar por ella.
Evitaría A Su Delta es buena, pero vence en 1 mes, antes de que se complete un alza "gradual de varios meses". Además tiene la Theta más agresiva (−0.10, 5 % diario del valor temporal).
C tiene poca Delta (0.20) y solo da ganancia sobre $125.
Si el alza fuera rápida y ocurriera dentro del primer mes, A sería mejor por su Delta.

#### **Escenario B:** anuncio en dos semanas, movimiento brusco, dirección desconocida
Compraría B. Al estar en el dinero, tiene la mayor sensibilidad de su Delta a los movimientos bruscos, es decir, la mayor convexidad (Gamma). Tiene además Vega considerable (0.25) y vence después del anuncio. Su prima de $8 es el costo máximo de pérdida.  
Evitaría A.	Se comporta como la acción (Delta 0.85), pero con Vega casi nula (0.05). Si el movimiento es a la baja, pierde casi como la acción, y si es al alza, no aprovecha la volatilidad. Su vencimiento a un mes es demasiado corto.  
Como el enunciado solo ofrece calls, una caída brusca genera pérdida. La estrategia ideal para un movimiento de dirección desconocida sería combinar una call con una put (straddle), pero eso queda fuera de las tres opciones.  
Riesgo de "volatility crush": tras el anuncio, la volatilidad implícita suele caer. Esto perjudicaría más a C (Vega 0.40) que a B.
C sería la alternativa si se busca la mayor exposición a volatilidad con pérdida acotada a $5.  

#### **Escenario C:** la acción se mantendrá muy estable

Si el precio no se mueve, la Delta no aporta nada. Solo importan Theta y Vega (la volatilidad tiende a bajar cuando el mercado está calmado).  
Comprar (menor pérdida)	A	Si la acción se queda en $100, A vale $10 al vencimiento. Pierde solo $2 (su valor temporal), es decir 17 % de la prima. Su Vega casi nula (0.05) la protege de una baja en la volatilidad.  
Evitaría	B	Al estar en el dinero, toda su prima ($8) es valor temporal y se pierde por completo si el precio no se mueve. También tiene Vega considerable (0.25).  
C también pierde toda su prima ($5) y es la más expuesta a una baja de volatilidad (Vega 0.40).  
La mejor decisión para esta visión sería vender una opción, no comprarla, porque Theta favorecería al vendedor. La candidata natural sería B, que tiene más valor temporal que cobrar. Esto implica riesgo si la acción se mueve.  
  
#### **Escenario D:** caída significativa del precio

Todas son calls, así que pierden valor con la caída. Se pierde más con más Delta.

Comprar (menor daño) C,	Delta de solo 0.20 y prima de $5, que es la pérdida máxima. Una caída de $10 restaría aproximadamente 0.20 × 10 = $2 por Delta.
Evitaría A,	Delta 0.85, la más alta. Una caída de $10 restaría aproximadamente 0.85 × 10 = $8.50 de una prima de $12.  
  
Con una visión bajista, ninguna call es un instrumento adecuado. Lo correcto sería una put, o vender una call.
Vender A generaría el mayor beneficio por Delta (0.85), con ganancia máxima igual a su prima ($12).  
  
#### **Escenario E:** el mercado subestima la incertidumbre futura

Esta visión equivale a esperar que la volatilidad implícita suba. Se busca la mayor Vega.

Compraría	C	Tiene la Vega más alta (0.40, o 8 % de su prima por punto de volatilidad). Su Theta es la más baja (−0.01), así que esperar a que el mercado corrija su visión cuesta poco. Su vencimiento a 1 año le da tiempo para que la incertidumbre se materialice.
Evitaría	A	Su Vega es la más baja (0.05, 0.42 % de la prima). Casi no se beneficia de un aumento de volatilidad, y tiene la Theta más agresiva.

```python
import numpy as np
import pandas as pd
from scipy.stats import norm
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Datos del enunciado
meses = ["Mayo", "Junio", "Julio", "Agosto", "Septiembre", "Octubre"]
S = np.array([100, 102, 101, 105, 103, 110], dtype=float)
sig = np.array([0.20, 0.25, 0.30, 0.28, 0.22, 0.20])
T_rest = np.array([6, 5, 4, 3, 2, 1]) / 12
 
r, K = 0.05, 100.0
V0 = 6.8887
delta, gamma, vega, theta = 0.5977, 0.02736, 0.2814, -8.116  # theta anualizada
dt = 1 / 12
```

```python
# 1) Aproximacion con Greeks (fijos en el dia inicial, recursiva)
#    Vega es "por 1 punto de volatilidad" => d_sigma se expresa en puntos (p.p.)

dS = np.r_[np.nan, np.diff(S)]
dsig_pp = np.r_[np.nan, np.diff(sig) * 100]   # en puntos porcentuales
 
V_greeks = [V0]
for i in range(1, len(S)):
    v = (V_greeks[-1] + delta * dS[i] + 0.5 * gamma * dS[i] ** 2
         + vega * dsig_pp[i] + theta * dt)
    V_greeks.append(v)
V_greeks = np.array(V_greeks)
 
# desglose de cada termino
desglose = pd.DataFrame({
    "Mes": meses,
    "Delta*dS": delta * dS,
    "0.5*Gamma*dS^2": 0.5 * gamma * dS ** 2,
    "Vega*dsigma": vega * dsig_pp,
    "Theta*dt": [np.nan] + [theta * dt] * 5,
})
```

```python
# 2) Black-Scholes

def bs_call(S0, K, r, s, T):
    d1 = (np.log(S0 / K) + (r + 0.5 * s ** 2) * T) / (s * np.sqrt(T))
    d2 = d1 - s * np.sqrt(T)
    return S0 * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)
 
def mc_call(S0, K, r, s, T, n=10000, seed=42):
    rng = np.random.default_rng(seed)
    z = rng.standard_normal(n // 2)
    z = np.r_[z, -z]                      # variables antiteticas
    ST = S0 * np.exp((r - 0.5 * s ** 2) * T + s * np.sqrt(T) * z)
    return np.exp(-r * T) * np.maximum(ST - K, 0).mean()
 
V_mc = np.array([mc_call(S[i], K, r, sig[i], T_rest[i]) for i in range(6)])
V_bs = np.array([bs_call(S[i], K, r, sig[i], T_rest[i]) for i in range(6)])
```

```python
# 3) Tablas
 
t1 = pd.DataFrame({"Mes": meses, "Precio": S, "Volatilidad": sig, "dS": dS,
                   "dsigma (p.p.)": dsig_pp, "Valor Greeks": V_greeks})
t2 = pd.DataFrame({"Mes": meses, "Precio": S, "Volatilidad": sig,
                   "Tiempo restante": [f"{int(t*12)}/12" for t in T_rest],
                   "BS Monte Carlo": V_mc, "BS formula cerrada": V_bs})
t3 = pd.DataFrame({"Mes": meses, "Greeks": V_greeks, "Black-Scholes (MC)": V_mc,
                   "Error absoluto": np.abs(V_greeks - V_mc)})
t3.loc[0, "Error absoluto"] = np.nan
 
pd.set_option("display.width", 200)
print(desglose.round(4).to_string(index=False), "\n")
print(t1.round(4).to_string(index=False), "\n")
print(t2.round(4).to_string(index=False), "\n")
print(t3.round(4).to_string(index=False))
```

```python
# 4) Grafica
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(meses, V_greeks, "o-", color="#d95f02", lw=2, label="Aproximacion con Greeks")
ax.plot(meses, V_mc, "s-", color="#1b6ca8", lw=2, label="Black-Scholes (Monte Carlo, 10,000)")
for i in range(6):
    ax.annotate(f"{V_greeks[i]:.2f}", (i, V_greeks[i]), textcoords="offset points",
                xytext=(0, -14), ha="center", fontsize=8, color="#d95f02")
    ax.annotate(f"{V_mc[i]:.2f}", (i, V_mc[i]), textcoords="offset points",
                xytext=(0, 8), ha="center", fontsize=8, color="#1b6ca8")
ax.set_title("Valor de la call: Greeks vs Black-Scholes")
ax.set_xlabel("Mes")
ax.set_ylabel("Valor de la opcion ($)")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
```

### **Preguntas**

#### **1. ¿Los greeks iniciales describen con precisión toda la trayectoria?**  
No. Solo funcionan cerca del punto inicial. El error crece hasta 1.07.  

#### **2. ¿En qué mes apareció el error más grande?**
Agosto (1.0726), seguido de cerca por Septiembre (1.0455).  

#### **3. ¿Qué factor parece contribuir más al error?**
La volatilidad y el tiempo (Vega y Theta fijas). Vega real baja al acercarse el vencimiento, y el desgaste de Theta se acelera, pero Theta fija lo trata como lineal.

#### **4. ¿Por qué los Greeks funcionan mejor para movimientos pequeños que para movimientos grandes?**
Los Greeks son derivadas (aproximación local, tipo Taylor). Al truncar, se ignoran términos de orden superior que crecen con dS. Además, los Greeks cambian cuando cambia el mercado.

#### **5. Si la trayectoria fuera mucho más volátil, ¿esperarías errores menores? ¿por qué?**
No, mayores. Más volatilidad implica movimientos más grandes y más se aleja la curva real de la aproximación.

#### **6. Concluye**
Los Greeks sirven para estimaciones rápidas de sensibilidad ante cambios pequeños. Para movimientos grandes o periodos largos hay que recalcular los Greeks o valuar con Black-Scholes o simulación.
